```python
# Convert Total Charges to numeric first
df["Total Charges"] = pd.to_numeric(df["Total Charges"], errors="coerce")

# Now check for negatives
print("Negative Monthly Charges:", (df["Monthly Charges"] < 0).sum())
print("Negative Total Charges:", (df["Total Charges"] < 0).sum())
```

```python
bins = [0, 6, 12, 24, 48, 72]
labels = [
    "0-6 Months",
    "7-12 Months",
    "13-24 Months",
    "25-48 Months",
    "49-72 Months"
]

df["tenure_group"] = pd.cut(
    df["Tenure Months"],   # <-- use the correct column name
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
import numpy as np

df["monthly_charge_group"] = pd.cut(
    df["Monthly Charges"],   # <-- use the correct column name
    bins=[0, 30, 60, 90, 120, np.inf],
    labels=[
        "Low",
        "Medium-Low",
        "Medium-High",
        "High",
        "Very High"
    ]
)
```

```python
def customer_segment(row):
    if row["Tenure Months"] <= 12 and row["Monthly Charges"] >= 70:
        return "High Risk"
    elif row["Tenure Months"] <= 24:
        return "Medium Risk"
    else:
        return "Low Risk"

# Apply to DataFrame
df["segment"] = df.apply(customer_segment, axis=1)
```

```python
def customer_segment(row):
    if row["Tenure Months"] <= 12 and row["Monthly Charges"] >= 70:
        return "High Risk"
    elif row["Tenure Months"] <= 24:
        return "Medium Risk"
    else:
        return "Low Risk"

df["risk_segment"] = df.apply(customer_segment, axis=1)
```

```python
import matplotlib.pyplot as plt

df["Churn Label"].value_counts().plot(
    kind="bar",
    color=["skyblue", "salmon"]  # optional styling
)

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Customers")
plt.show()
```

```python
churn_rate = (
    df["Churn Label"].eq("Yes").mean() * 100
)

print(f"Churn Rate: {churn_rate:.2f}%")
```

```python
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn Label"],
    normalize="index"
) * 100

contract_churn
```

```python
tenure_churn = (
    df.groupby("tenure_group", observed=True)["Churn Label"]
      .apply(lambda x: (x == "Yes").mean() * 100)
)

tenure_churn
```

```python
payment_churn = (
    df.groupby("Payment Method")["Churn Label"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
)

payment_churn
```

```python
import matplotlib.pyplot as plt

for churn_status, group in df.groupby("Churn Label"):
    plt.scatter(
        group["Tenure Months"],
        group["Monthly Charges"],
        label=churn_status,
        alpha=0.5
    )

plt.xlabel("Tenure (Months)")
plt.ylabel("Monthly Charges")
plt.title("Tenure vs Monthly Charges by Churn")
plt.legend()
plt.show()
```
